```python
import pandas as pd
```

```python
df=pd.read_csv("C:/Users/ramya/Downloads/amazon_sales_dataset.csv")
print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
# Check dataset information
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())

# Display first 5 records
df.head()
```

```python
# Convert order_date to datetime
df["order_date"] = pd.to_datetime(df["order_date"])

# Check the data type
print(df["order_date"].dtype)

# Display date range
print("Start Date:", df["order_date"].min())
print("End Date:", df["order_date"].max())
```

```python
# Check minimum values of numeric columns
print("Minimum Values:")
print(df[[
    "price",
    "discount_percent",
    "quantity_sold",
    "rating",
    "review_count",
    "discounted_price",
    "total_revenue"
]].min())
```

```python
# Check maximum values of numeric columns
print("Maximum Values:")
print(df[[
    "price",
    "discount_percent",
    "quantity_sold",
    "rating",
    "review_count",
    "discounted_price",
    "total_revenue"
]].max())
```

```python
# Standardize text columns
text_columns = [
    "product_category",
    "customer_region",
    "payment_method"
]

for col in text_columns:
    df[col] = df[col].str.title()

# Check the unique values
for col in text_columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
# Create date-based columns
df["Year"] = df["order_date"].dt.year
df["Month"] = df["order_date"].dt.month

# Check the new columns
print(df[["order_date", "Year", "Month"]].head())

print("\nYears:", df["Year"].unique())
print("Months:", df["Month"].unique())
```

```python
# Calculate expected revenue
df["Calculated_Revenue"] = (
    df["discounted_price"] * df["quantity_sold"]
)

# Check difference
df["Revenue_Difference"] = (
    df["total_revenue"] - df["Calculated_Revenue"]
)

print("Maximum Revenue Difference:",
      df["Revenue_Difference"].abs().max())

print("\nSample comparison:")
print(df[[
    "discounted_price",
    "quantity_sold",
    "total_revenue",
    "Calculated_Revenue",
    "Revenue_Difference"
]].head())
```

```python
# Remove temporary columns
df.drop(
    columns=["Calculated_Revenue", "Revenue_Difference"],
    inplace=True
)

# Check final columns
print("Columns:", df.columns.tolist())
print("Shape:", df.shape)
```

```python
# Final data quality check
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
# Save the cleaned Sales dataset
output_path = "C:/Users/ramya/Downloads/amazon_sales_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned Sales dataset saved successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```
