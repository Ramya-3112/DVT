```python
import matplotlib.pyplot as plt
df=pd.read_csv("C:/Users/ramya/Downloads/archive (4).zip")
```

```python
category_count=df["product_category"].value_counts()
category_count.plot(kind="bar",figsize=(10,5))
plt.title("Number of Products by Category")
plt.xlabel("Product Category")
plt.ylabel("Count")
plt.show()
```

```python
payment_count = df["payment_method"].value_counts()

payment_count.plot(kind="bar", figsize=(8,5))
plt.title("Payment Methods Used")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.show()
```

```python
df["rating"].plot(kind="hist", bins=10)
plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.show()
```
