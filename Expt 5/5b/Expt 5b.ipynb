```python
import matplotlib.pyplot as plt          

# sales by category
category_sales = df.groupby(
    "product_category"
)["total_revenue"].sum()                 
plt.figure(figsize=(8,5))                
plt.bar(
    category_sales.index,
    category_sales.values
)                                       
plt.title("Total Revenue by Category")   
plt.xlabel("Product Category")          
plt.ylabel("Total Revenue")              
plt.xticks(rotation=45)                  
plt.show()                               
```

```python
# Pie chart - Payment methods
payment_count = df["payment_method"].value_counts() 

plt.figure(figsize=(7,7))                            # Set chart size

plt.pie(
    payment_count.values,
    labels=payment_count.index,
    autopct="%1.1f%%"
)                                                    # Create pie chart

plt.title("Payment Method Distribution")             # Add chart title

plt.show()                                           # Display pie chart
```

```python
region_revenue = df.groupby(
    "customer_region"
)["total_revenue"].sum()                 # Calculate regional revenue

plt.figure(figsize=(10,5))               # Set chart size

plt.bar(
    region_revenue.index,
    region_revenue.values
)                                        # Create bar chart

plt.title("Total Revenue by Region")     # Add chart title

plt.xlabel("Customer Region")            # Label x axis

plt.ylabel("Total Revenue")              # Label y axis

plt.xticks(rotation=45)                  # Rotate region labels

plt.show()                               # Display bar chart
```

```python
plt.figure(figsize=(8,5))                # Set chart size

plt.scatter(
    df["price"],
    df["rating"],
    s=10,
    alpha=0.5
)                                        # Create scatter plot

plt.title("Price vs Rating")             # Add chart title

plt.xlabel("Price")                      # Label x axis

plt.ylabel("Rating")                     # Label y axis

plt.show()                               # Display scatter plot
```

```python
plt.figure(figsize=(10,5))               # Set chart size

plt.plot(
    df["order_id"].head(20),
    df["total_revenue"].head(20),
    marker="o"
)                                        # Create line chart

plt.title("Revenue by Order")            # Add chart title

plt.xlabel("Order ID")                   # Label x axis

plt.ylabel("Total Revenue")              # Label y axis

plt.grid(True)                           # Add grid lines

plt.show()                               # Display line chart
```

```python
import matplotlib.pyplot as plt          # Import plotting library

plt.figure(figsize=(8,5))                # Set chart size

plt.boxplot(df["total_revenue"])         # Create revenue box plot

plt.title("Outlier Detection in Total Revenue") # Add chart title

plt.ylabel("Total Revenue")              # Label y axis

plt.show()                               # Display box plot
```

```python
import matplotlib.pyplot as plt          # Import plotting library

plt.figure(figsize=(8,5))                # Set chart size

plt.hist(df["price"], bins=10)           # Create price histogram

plt.title("Price Distribution")          # Add chart title

plt.xlabel("Price")                      # Label x axis

plt.ylabel("Frequency")                  # Label y axis

plt.show()                               # Display histogram
```
