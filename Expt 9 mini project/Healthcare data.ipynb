```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicates removed.")
print("Current dataset shape:", df.shape)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
)

print(df.columns.tolist())
```

```python
df["Date_of_Admission"] = pd.to_datetime(
    df["Date_of_Admission"],
    errors="coerce"
)

df["Discharge_Date"] = pd.to_datetime(
    df["Discharge_Date"],
    errors="coerce"
)

print(df[["Date_of_Admission", "Discharge_Date"]].dtypes)
```

```python
print(df[["Age", "Billing_Amount", "Room_Number"]].describe())
```

```python
# Check negative billing amounts
negative_billing = (df["Billing_Amount"] < 0).sum()

print("Number of negative billing amounts:", negative_billing)
```

```python
negative_bills = df[df["Billing_Amount"] < 0]

print("Negative billing records:")
print(negative_bills[[
    "Name",
    "Age",
    "Medical_Condition",
    "Billing_Amount",
    "Admission_Type"
]].head(20))
```

```python
df = df[df["Billing_Amount"] >= 0].copy()

print("Negative billing records removed.")
print("Current dataset shape:", df.shape)
```

```python
print("Negative Age values:", (df["Age"] < 0).sum())
print("Negative Room Number values:", (df["Room_Number"] < 0).sum())
print("Negative Length of Stay:", (df["Length_of_Stay"] < 0).sum())
```

```python
df["Length_of_Stay"] = (
    df["Discharge_Date"] - df["Date_of_Admission"]
).dt.days

df[[
    "Date_of_Admission",
    "Discharge_Date",
    "Length_of_Stay"
]].head()
```

```python
print(
    "Invalid Length of Stay:",
    (df["Length_of_Stay"] < 0).sum()
)
```

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 18, 30, 45, 60, 75, 100],
    labels=[
        "0-18",
        "19-30",
        "31-45",
        "46-60",
        "61-75",
        "76-100"
    ]
)

df[["Age", "Age_Group"]].head()
```

```python
print("Gender:")
print(df["Gender"].value_counts())

print("\nMedical Conditions:")
print(df["Medical_Condition"].value_counts())

print("\nAdmission Types:")
print(df["Admission_Type"].value_counts())

print("\nTest Results:")
print(df["Test_Results"].value_counts())
```

```python
df["Name"] = df["Name"].str.title()

print(df["Name"].head(10))
```

```python
text_columns = [
    "Gender",
    "Blood_Type",
    "Medical_Condition",
    "Admission_Type",
    "Test_Results"
]

for col in text_columns:
    df[col] = df[col].str.title()

print(df[text_columns].head())
```

```python
print("Admission date range:")
print(df["Date_of_Admission"].min(), "to", df["Date_of_Admission"].max())

print("\nDischarge date range:")
print(df["Discharge_Date"].min(), "to", df["Discharge_Date"].max())
```

```python
print("Final dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values after preprocessing:")
print(df.isnull().sum())
```
