```python
high_price = df[df["price"] > 490]
print(high_price)
```

```python
high_rating = df[df["rating"] == 5]
print(high_rating)
```

```python
high_revenue = df[df["total_revenue"] > 2400]

print(high_revenue)
```

```python
electronics_products = df[df["product_category"] == "Electronics"]
print(electronics_products)
```

```python
sorted_df = df.sort_values(
    by="rating",
    ascending=False
)
print(sorted_df)
```

```python
import matplotlib.pyplot as plt
sorted_df = df.sort_values(by="rating", ascending=False)
plt.figure(figsize=(10,5))
plt.bar(sorted_df["product_id"].head(10).astype(str),
        sorted_df["rating"].head(10))
plt.title("Top 10 Products by Rating")
plt.xlabel("Product ID")
plt.ylabel("Rating")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
sorted_df = df.sort_values(by="total_revenue", ascending=False)
plt.figure(figsize=(10,5))
plt.bar(sorted_df["product_id"].head(10).astype(str),
        sorted_df["total_revenue"].head(10))
plt.title("Top 10 Products by Total Revenue")
plt.xlabel("Product ID")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
print("Mean =", df["price"].mean())
print("Median =", df["price"].median())
print("Mode =", df["price"].mode())
print("Standard Deviation =", df["price"].std())
```

```python
df.describe()
```

```python
category_avg = df.groupby(
    "product_category"
)["rating"].mean()
print(category_avg)
```

```python
category_avg = df.groupby("product_category")["rating"].mean()
plt.figure(figsize=(8,5))
plt.plot(category_avg.index, category_avg.values, marker='o', linewidth=2)
plt.title("Average Rating by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
corr = df[[
    "rating",
    "review_count"
]].corr()
print(corr)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["review_count"],
            df["rating"],
            s=5,
            alpha=0.3)

plt.title("Rating vs Review Count")
plt.xlabel("Review Count")
plt.ylabel("Rating")

plt.show()
```
